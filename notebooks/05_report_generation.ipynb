## Hallucination-Resistant Medical Report Generation Using Generative AI

### Research Objectives:
1. Construct structured input conditioning combining visual findings, probabilities, and retrieved medical context.
2. Synthesize complete, standardized radiology reports with Examination, Findings, and Impression sections.
3. Compare baseline generation (findings only) against RAG-enhanced generation.

```python
# 1. Module Imports and Generator Initialization
import sys
from pathlib import Path
for p in [Path.cwd(), Path.cwd().parent, Path('/kaggle/working/hallucination_medical_ai')]:
    if (p / 'src').exists() and str(p) not in sys.path:
        sys.path.insert(0, str(p))

from src.config import get_config
from src.generation.report_generator import MedicalReportGenerator

cfg = get_config()
generator = MedicalReportGenerator(cfg)
print('Report generator initialized successfully.')
```

### 2. Format Structured Findings Input


```python
sample_visual_findings = {
    'Cardiomegaly': 0.84,
    'Pleural Effusion': 0.12,
    'Pneumothorax': 0.03,
    'Consolidation': 0.08,
    'Atelectasis': 0.45,
    'Edema': 0.10,
    'Lung Opacity': 0.15,
    'Enlarged Cardiomediastinum': 0.65,
    'Lung Lesion': 0.05,
    'No Finding': 0.05
}

retrieved_context = [
    'Cardiomegaly is defined as cardiothoracic ratio exceeding 50 percent on standard PA radiograph.',
    'Basilar atelectasis can manifest as linear horizontal opacities in dependent lung bases.'
]

generation_input = {
    'visual_findings': sample_visual_findings,
    'retrieved_context': retrieved_context,
    'image_view': 'PA',
    'patient_age': '62',
    'patient_sex': 'F'
}
```

### 3. Generate Structured Radiology Report


```python
report = generator.generate_report(generation_input)
print('--- GENERATED STRUCTURED REPORT ---')
print(report.to_text())
```

### 4. Verify Export Formats (JSON and Readable Text)


```python
report_json = report.to_dict()
print('Exported JSON keys:', list(report_json.keys()))
print('Impression:', report_json.get('impression'))
```
